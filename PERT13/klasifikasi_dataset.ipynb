# Identifikasi Dataset Klasifikasi
Satu dataset untuk tiap jenis klasifikasi: **Biner**, **Multikelas**, **Multilabel**.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer, load_iris, fetch_openml
```

## 1. Klasifikasi Biner: Breast Cancer Wisconsin
**Sumber:** UCI Machine Learning Repository, dimuat lewat `sklearn.datasets.load_breast_cancer`.

```python
d1 = load_breast_cancer(as_frame=True)
df1 = d1.frame                      # ubah ke DataFrame

X1 = df1.drop(columns="target")     # data fitur
y1 = df1["target"]                  # data label

print("Jenis data      : Tabular numerik (bukan teks / image)")
print("Jumlah data     :", df1.shape[0])
print("Jumlah variabel :", df1.shape[1], "(", X1.shape[1], "fitur + 1 target )")
print("Kolom label     : target")
print("Nama kelas      :", dict(enumerate(d1.target_names)))
print("Jumlah kelas    :", y1.nunique())
print("Kesimpulan      : Klasifikasi", "BINER" if y1.nunique() == 2 else "MULTIKELAS")
df1.head()
```

## 2. Klasifikasi Multikelas: Iris
**Sumber:** UCI Machine Learning Repository (R. A. Fisher, 1936), dimuat lewat `sklearn.datasets.load_iris`.

```python
d2 = load_iris(as_frame=True)
df2 = d2.frame

X2 = df2.drop(columns="target")
y2 = df2["target"]

print("Jenis data      : Tabular numerik (bukan teks / image)")
print("Jumlah data     :", df2.shape[0])
print("Jumlah variabel :", df2.shape[1], "(", X2.shape[1], "fitur + 1 target )")
print("Kolom label     : target")
print("Nama kelas      :", dict(enumerate(d2.target_names)))
print("Jumlah kelas    :", y2.nunique())
print("Kesimpulan      : Klasifikasi", "BINER" if y2.nunique() == 2 else "MULTIKELAS")
df2.head()
```

## 3. Klasifikasi Multilabel: Dataset Sintetis
**Sumber:** dibangkitkan dengan `sklearn.datasets.make_multilabel_classification` (`random_state=42`), disimpan di `multilabel_sintetis.csv`. Satu data bisa memiliki beberapa label sekaligus.

```python
df3 = pd.read_csv("multilabel_sintetis.csv")

# kolom label = label_1..label_5, sisanya fitur
label_cols = [c for c in df3.columns if str(c).startswith("label_")]
X3 = df3.drop(columns=label_cols)
Y3 = df3[label_cols].astype(int)

print("Jenis data      : Tabular numerik (bukan teks / image)")
print("Jumlah data     :", df3.shape[0])
print("Jumlah variabel :", df3.shape[1], "(", X3.shape[1], "fitur +", len(label_cols), "label )")
print("Kolom label     :", label_cols)
print("Rata-rata label per data:", round(Y3.sum(axis=1).mean(), 2))
print("Kesimpulan      : Klasifikasi", "MULTILABEL" if Y3.sum(axis=1).max() > 1 else "MULTIKELAS")
df3.head()
```
