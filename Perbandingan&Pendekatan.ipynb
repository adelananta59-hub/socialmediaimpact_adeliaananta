# Penjelasan MAE, MSE dan R^2 Pada Algoritma Regrensi

Disini kita harus mencari tau mengenai refrensi artikel serta refrensi teori tentang MAE, MSE dan R^2 apakah lebih bagus lebih dekat dengan angka 1 atau 0, 
serta R^2 juga apakah lebih bagus mendekati angka kurang dari 1 dan angka mendekati 0. Mari kita bahas dengan beberapa refrensi artikel serta teori dibawah ini
Pertama kita ketahui dulu bahwa MAE serta MSE makin mendekati 0 maka semakin bagus, sebenarnya angka 1 bukanlah patokannya. MAE punya satuan yang sama dengan target, misal harga rumah
dalam dolar, MAE 20.000 bisa tergolong bagus. Sementara R^2 semakin baik jika nilainya mendekati 1. Penjelasan ini sesuai dengan teori evaluasi model regresi dalam artikel Evaluation metrics and statistical tests for machine learning yang diterbitkan di jurnal Scientific Reports pada tahun 2024

# Cara membaca nilai evaluasi regresi

1.) MAE = Mendekati 0	Rata-rata selisih absolut prediksi dengan nilai sebenarnya semakin kecil.
2.) MSE = (Mean Squared Error)	Mendekati 0	Rata-rata kesalahan kuadrat semakin kecil. Kesalahan besar mendapat penalti lebih besar.
3.) R² = (R-squared)	Mendekati 1	Model semakin baik dalam menjelaskan variasi target pada data evaluasi.

Apakah nilainya harus kurang dari 1?
- MAE: Tidak harus kurang dari 1. Nilainya bergantung pada skala target. Misalnya, MAE sebesar 5 bisa sangat baik jika memprediksi harga dalam jutaan rupiah, tetapi bisa buruk jika targetnya hanya berkisar 0–10.
- MSE: Tidak harus kurang dari 1. Nilainya juga bergantung pada skala target dan dikuadratkan, sehingga bisa jauh lebih besar daripada MAE.
- R²: Umumnya berada antara 0 dan 1 untuk model dengan hasil evaluasi yang baik atau setidaknya tidak lebih buruk daripada prediksi menggunakan rata-rata target. Namun, R² juga dapat bernilai negatif jika model lebih buruk daripada prediksi rata-rata pada data pengujian.
Kesimpulan: MAE dan MSE bukan dinilai berdasarkan apakah nilainya di bawah 1, tetapi berdasarkan seberapa kecil kesalahannya. R² dinilai berdasarkan seberapa tinggi nilainya, dengan 1 sebagai nilai ideal.

# Contoh kodenyauntuk menghasilkan prediksi y_pred


from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Menghitung nilai evaluasi model regresi
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# Menampilkan hasil evaluasi
print("Hasil Evaluasi Model Regresi")
print("----------------------------")
print(f"MAE : {mae:.4f}")
print(f"MSE : {mse:.4f}")
print(f"R2  : {r2:.4f}")

# Interpretasi sederhana
if mae >= 0 and mse >= 0:
    print("\nMAE dan MSE: semakin kecil nilainya, semakin baik.")
    
if r2 >= 0.9:
    print("R2: sangat tinggi, mendekati 1.")
elif r2 >= 0.7:
    print("R2: relatif tinggi.")
elif r2 >= 0:
    print("R2: model masih perlu dievaluasi.")
else:
    print("R2: model lebih buruk daripada prediksi rata-rata pada data uji.")


Interpretasinya:
1. MAE = 2,5: rata-rata kesalahan absolut prediksi adalah 2,5 satuan target.
2. MSE = 10: rata-rata kuadrat kesalahan prediksi adalah 10 satuan kuadrat.
3. R² = 0,92: model menjelaskan sekitar 92% variasi target relatif terhadap model pembanding yang memprediksi rata-rata target pada data uji. Ini menunjukkan hasil yang baik, tetapi tetap perlu diperiksa terhadap kemungkinan overfitting dan kualitas data.
Angka tersebut hanya contoh, bukan hasil pengujian dataset Anda. Nilai MAE dan MSE tidak dapat ditafsirkan secara terpisah dari skala target.

# Evaluasi Kinerja Algoritma Regresi

Evaluasi kinerja algoritma regresi dapat dilakukan menggunakan Mean Absolute Error (MAE), Mean Squared Error (MSE), dan koefisien determinasi (R²). MAE mengukur rata-rata selisih absolut antara nilai aktual dan nilai prediksi, sedangkan MSE mengukur rata-rata kuadrat kesalahan prediksi. Nilai MAE dan MSE yang semakin kecil menunjukkan kesalahan prediksi yang semakin rendah. Namun, kedua metrik tersebut tidak memiliki batas ideal universal sebesar 1 karena nilainya dipengaruhi oleh skala variabel target.

Sementara itu, R² digunakan untuk mengevaluasi kemampuan model dalam menjelaskan variasi variabel target. Nilai R² yang mendekati 1 menunjukkan bahwa model mampu menjelaskan variasi target dengan baik. Nilai R² sebesar 0 menunjukkan bahwa model tidak memberikan peningkatan dibandingkan prediksi menggunakan rata-rata target pada data evaluasi, sedangkan nilai negatif menunjukkan kinerja yang lebih buruk daripada pembanding tersebut. Oleh karena itu, evaluasi model regresi sebaiknya mempertimbangkan MAE, MSE, dan R² secara bersamaan serta memperhatikan karakteristik dataset.

Penjelasan ini didukung oleh Rainio et al. (2024) dan artikel Common statistical concepts in the supervised Machine Learning arena (2023).
